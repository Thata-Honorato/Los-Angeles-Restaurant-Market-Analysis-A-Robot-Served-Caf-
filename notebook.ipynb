```python
# Bibliotecas principais para análise e visualização
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Estilo para os gráficos
sns.set(style='whitegrid')
```

```python

# Carregando o arquivo
file_path = '/datasets/rest_data_us_upd.csv'
df = pd.read_csv(file_path)

# Visualizando as 5 primeiras linhas
df.head()

```

```python
# Visualizando informações gerais
df.info()

# Verificando estatísticas descritivas
df.describe(include='all')
```

```python
# Verificando se há linhas duplicadas
duplicatas = df.duplicated().sum()
print(f"Número de linhas duplicadas: {duplicatas}")

# Removendo duplicatas, se houver
df = df.drop_duplicates().reset_index(drop=True)
```

```python
# Verificação final após limpeza
df.info()
df.head()
```

```python

# Ver quantos valores ausentes
print(df['chain'].isna().sum())

# Substituir os valores ausentes por 'False'
df['chain'] = df['chain'].fillna(False)

```


<div class="alert alert-success">
<strong>Comentário do revisor v1</strong>

Excelente trabalho checando a consistência dos dados.

Sempre importante para garantir uma análise robusta!
</div>

<div class="alert alert-success">

```python
# Distribuição: faz parte de rede ou não
df['chain'].value_counts().plot(kind='bar', title='Estabelecimentos que fazem parte de redes')
plt.xticks([0, 1], ['Independentes', 'Redes'], rotation=0)
plt.ylabel('Quantidade')
plt.show()
```

<div class="alert alert-success">

```python
# Histograma da coluna 'number'
plt.figure(figsize=(10, 4))
sns.histplot(df['number'], bins=50, kde=True)
plt.title('Distribuição do número de funcionários')
plt.xlabel('Número de funcionários')
plt.ylabel('Frequência')
plt.show()
```

```python
# Boxplot para identificar outliers
plt.figure(figsize=(8, 2))
sns.boxplot(x=df['number'])
plt.title('Boxplot do número de funcionários')
plt.show()
```

```python
# Frequência absoluta
type_counts = df['object_type'].value_counts()
print(type_counts)

# Frequência relativa (%)
type_percent = df['object_type'].value_counts(normalize=True) * 100
print(type_percent)

# Para entender qual o tipo de restaurante mais comum
```

```python
# Agrupar por 'chain' e calcular média de funcionários
chain_group = df.groupby('chain')['number'].agg(['mean', 'median', 'count'])
print(chain_group)
```

```python
# Boxplot para comparar número de funcionários por tipo (rede vs. independente)
sns.boxplot(data=df, x='chain', y='number')
plt.xticks([0, 1], ['Independentes', 'Redes'])
plt.title('Funcionários: Redes vs. Independentes')
plt.xlabel('Faz parte de rede?')
plt.ylabel('Número de funcionários')
plt.show()

# confirmar se as redes têm mais funcionários, o que pode indicar uma operação mais estruturada e escalável.

```

```python
# Quantos endereços únicos
print(f"Endereços únicos: {df['address'].nunique()}")

# Endereços duplicados
duplicated_addresses = df[df.duplicated(subset='address', keep=False)]
print(f"Endereços duplicados:\n{duplicated_addresses.head()}")
```

#### Análises iniciais: 


O mercado é dominado por restaurantes convencionais (75%).

Cafeterias representam apenas 4,5% do total — isso pode indicar:

* Uma oportunidade de nicho, caso a demanda por cafés seja alta.

* Ou que o mercado não comporta muitos cafés.

O fato de bares, padarias e pizzarias também representarem pouco pode reforçar o domínio dos restaurantes tradicionais.



```python
# extrair e analisar somente os registros do tipo "Cafe"
cafes_df = df[df['object_type'] == 'Cafe']

#Quantidade total de cafés
total_cafes = cafes_df.shape[0]
print(f'Total de cafés: {total_cafes}')
```

```python
# Proporção de cafés que fazem parte de rede
rede = cafes_df['chain'].value_counts()
print('\nCafés em rede (True/False):')
print(rede)
print('\nProporção (%):')
print((rede / total_cafes * 100).round(2))
```

```python
# Estatísticas de funcionários por café
print('\nNúmero de funcionários por café (estatísticas):')
print(cafes_df['number'].describe())
```

```python
# Endereços duplicados entre cafés
duplicados_cafe = cafes_df[cafes_df.duplicated(subset='address', keep=False)]
print(f'\nCafés com endereço duplicado: {duplicados_cafe.shape[0]}')
print(duplicados_cafe[['object_name', 'address', 'chain', 'number']].head())
```

--------
Há forte presença de redes na cidade. 
Esses dados sugerem que o mercado é altamente competitivo e consolidado por grandes marcas, o que pode ser desafiador para novos entrantes, a menos que tragam algo realmente inovador — como o conceito de garçons robôs.


A presença de cafés com até 143 funcionários sugere operações grandes, provavelmente em aeroportos ou centros corporativos.
Como 75% dos cafés têm até 34 funcionários, a automação (como robôs) poderia ser economicamente interessante para reduzir custos operacionais em parte desses estabelecimentos.


Os endereços duplicados podem estar atrelados a areroportos, praças de alimentação ou shoppings, todos compartilham do mesmo espaço.

```python
# Extrair a última parte do endereço (nome da rua)
df_cafes = df[df['object_type'] == 'Cafe'].copy()
df_cafes['rua'] = df_cafes['address'].str.extract(r'(\b[A-Z ]+)$')

```

```python


# Contar cafés por rua
cafes_por_rua = df_cafes['rua'].value_counts().head(10)
print(cafes_por_rua)

```

```python
# Salvar o gráfico em arquivo PNG
plt.figure(figsize=(10,6))
cafes_por_rua.plot(kind='bar')
plt.title('Top 10 ruas com mais cafés em Los Angeles')
plt.xlabel('Rua')
plt.ylabel('Número de cafés')
plt.xticks(rotation=45)
plt.tight_layout()

# Salvar o gráfico em um arquivo
plt.savefig('top10_cafes_por_rua.png')

plt.show()

# Exibir link para download
from IPython.display import FileLink
display(FileLink('top10_cafes_por_rua.png'))
```

------

As ruas com maior número de cafés são W Sunset Blvd (21 cafés) e W Pico Blvd (11 cafés), seguidas por zonas comerciais clássicas como Santa Monica Blvd, Hollywood Blvd e Wilshire Blvd.

Isso indica alta demanda nessas regiões, mas também alta concorrência.

Uma estratégia interessante seria investigar ruas adjacentes a essas vias com tráfego elevado, porém menor densidade de cafés — favorecendo a inovação do modelo com garçons robôs e destacando-se da concorrência tradicional.

```python
# Extrair o nome da rua da coluna 'address'
df['rua'] = df['address'].str.extract(r'(\b[\w\s]+)$')
```

```python
# Contar quantos restaurantes tem em cada rua
rua_counts = df['rua'].value_counts()

print(rua_counts.head(10))
```

```python
# Filtrar as ruas que têm 5 ou mais restaurantes
top_ruas = rua_counts[rua_counts >= 5].index

# Filtrar os dados do DataFrame para essas ruas
rest_top_ruas = df[df['rua'].isin(top_ruas)]
```

```python
# Visualizar a distribuição do número de assentos por rua usando boxplots
plt.figure(figsize=(14, 8))
sns.boxplot(data=rest_top_ruas, x='rua', y='number')
plt.xticks(rotation=45)
plt.title('Distribuição do número de assentos nas ruas com muitos restaurantes')
plt.ylabel('Número de assentos')
plt.xlabel('Rua')
plt.show()
```

```python
# Contar quantas ruas têm apenas um restaurante
ruas_com_um_restaurante = (rua_counts == 1).sum()

print(f"Número de ruas com apenas um restaurante: {ruas_com_um_restaurante}")
```

```python
# Contar quantos restaurantes tem em cada rua
rua_counts = df['rua'].value_counts()
```

```python
# Filtrar as ruas que têm muitos restaurantes (5 ou mais)
top_ruas = rua_counts[rua_counts >= 5].index

# Filtrar os dados para essas ruas
rest_top_ruas = df[df['rua'].isin(top_ruas)]
```

```python

# Visualizar a distribuição do número de assentos nas ruas com muitos restaurantes usando boxplot
plt.figure(figsize=(14, 8))
sns.boxplot(data=rest_top_ruas, x='rua', y='number')
plt.xticks(rotation=45)
plt.title('Distribuição do número de assentos nas ruas com muitos restaurantes')
plt.xlabel('Rua')
plt.ylabel('Número de assentos')
plt.show()
```

```python
# Contar quantos estabelecimentos por tipo
tipo_counts = df['object_type'].value_counts().reset_index()
tipo_counts.columns = ['object_type', 'count']
```

```python
# Calcular a média de assentos por tipo
tipo_mean_seats = df.groupby('object_type')['number'].mean().reset_index()
```

```python
# Juntar os dois dataframes
tipo_stats = tipo_counts.merge(tipo_mean_seats, on='object_type')

tipo_stats
```

```python
fig, ax1 = plt.subplots(figsize=(10,6))

# Gráfico de barras: quantidade por tipo
sns.barplot(data=tipo_stats, x='object_type', y='count', ax=ax1, color='skyblue')
ax1.set_ylabel('Quantidade de Estabelecimentos')
ax1.set_xlabel('Tipo de Estabelecimento')
ax1.set_title('Quantidade e Média de Assentos por Tipo de Estabelecimento')


# Criar segundo eixo y para a média de assentos
ax2 = ax1.twinx()
sns.lineplot(data=tipo_stats, x='object_type', y='number', marker='o', ax=ax2, color='red')
ax2.set_ylabel('Média de Assentos')

plt.show()
```

```python
# Contagem dos tipos de estabelecimentos por rede e não rede
tipo_chain = df.groupby(['chain', 'object_type'])['object_name'].count().reset_index()

# Filtrar apenas redes (chain = True)
tipo_chain_true = tipo_chain[tipo_chain['chain'] == True].sort_values(by='object_name', ascending=False)

# Top 10 tipos mais comuns entre redes
plt.figure(figsize=(10, 6))
sns.barplot(x='object_name', y='object_type', data=tipo_chain_true.head(10), palette='viridis')
plt.title('Top 10 tipos de estabelecimento em redes')
plt.xlabel('Quantidade de estabelecimentos')
plt.ylabel('Tipo de Estabelecimento')
plt.show()


```

```python
# Cálculo de estatísticas de assentos por rede e não rede
assentos_chain = df.groupby('chain')['number'].agg(['mean', 'median', 'count']).reset_index()

# Boxplot para visualizar a distribuição
plt.figure(figsize=(8,6))
sns.boxplot(x='chain', y='number', data=df, palette='Set2')
plt.title('Distribuição do número de assentos: rede vs não rede')
plt.xlabel('Rede')
plt.ylabel('Número de assentos')
plt.xticks([0,1], ['Não Rede', 'Rede'])
plt.show()

assentos_chain
```

```python
plt.figure(figsize=(14, 8))

sns.boxplot(data=df, x='object_type', y='number', hue='chain', palette='Set2')

plt.title('Distribuição do Número de Assentos por Tipo de Estabelecimento e Rede')
plt.xlabel('Tipo de Estabelecimento')
plt.ylabel('Número de Assentos')
plt.xticks(rotation=45)
plt.legend(title='Estabelecimento em Rede (chain)')

# Salvar figura como PNG (ou JPG)
plt.savefig('grafico_assentos_tipo_rede.png', dpi=300, bbox_inches='tight')

plt.show()

from IPython.display import FileLink

# Mostra o link para baixar o arquivo
display(FileLink('grafico_assentos_tipo_rede.png'))
```

Isso sugere que redes priorizam muitos pontos pequenos para cobertura geográfica, enquanto independentes optam por menos unidades, porém, maiores.

Redes tendem a ser mais comuns em restaurantes e fast food. Além disso, observa-se que estabelecimentos em rede geralmente possuem uma maior variação no número de assentos, indicando presença tanto de unidades pequenas quanto grandes. Cafés apresentam menor número médio de assentos, mas têm participação significativa em redes, sugerindo potencial para expansão em formato de rede com unidades de tamanho mais uniforme.
